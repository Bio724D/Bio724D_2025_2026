### Assignment

Python has several powerful idioms for assignment that can simplify code and make it easier to read. When using these idioms, keep in mind that the values on the left hand side (LHS) of the statment is/are assigned the identifiers (variable names) on the right hand side (RHS). 

**Simultaneous assignment** simplifies code. 

```python
# simultaneous assignment of multiple values to multiple variables
a, b, c = 'mouse', 'leopard', 'orca'   # assigns in order given; balance items on RHS and LHS!
```

```python
# simultaneous assignment of a single value to multiple variables
a = b = c = 'mouse'				       # right-most becomes RHS, others are LHS
```

```python
c
```

**Direct swapping** eliminates the need for a temporary variable. Suppose you want to exchange the values assigned to a and b:

```python
# initialize values
a = 'spam'
b = 'eggs'
```

```python
# traditional way to swap values
temp = a
a = b
b = temp
```

```python
# direct swapping (not available in most languages)
a, b = b, a
```

```python
print('a =', a, 'b =', b)
```

**Unpacking** is a powerful Python feature that greatly simplifies the process of extracting items from a list or tuple. It is particularly useful when you don't know the length of container.

The following examples query this object: `mammals = ['mouse', 'fruit bat', 'chimpanzee', 'leopard']`

```python
# assign without the need for indexing
i, j, k, l = mammals               # items assigned in order
```

```python
# unpack some items but not all
i, _, j, _ = mammals               # use _ to ignore a single item
```

```python
# unpack some items into a list, others to single values
i, *j  = mammals                  	# use * to assign other items to a list
```

```python
# unpack just the first or last items
# use *_ to ignore the other items
i, *_  = mammals                   # assigns first item only
*_, j  = mammals                   # assigns last item only
i, *_, j  = mammals                # assigns first and last items only
```

```python
print('i =', i, 'j=', j)
```

**Augmented assignment**, also known as compound assignment, combines an operation with assignment in a single step. This has two benefits: it simplifies code, making it easier to read, and it results in slightly faster execution. The following examples use `+=`, `-=` and `*=`, respectively agumented assignment operators for addition, subtraction, and multiplication. Python offers many other augmented assignment operators, but most are not commonly encountered.  

```python
# traditional way to update a tally
tally = 0
for i in range(10):
    tally = tally + 1
```

```python
# update a tally with agumented assignment
tally = 0
for i in range(10):
    tally += 1
```

```python
# not only in loops; continuously updating a tally
tally = 0
tally += 23
tally *= 2
tally -= 4
```

```python
tally
```

In this particular case, for example, we need to test whether the item is in the list before asking for it's position.  

```python
# traditional approach using an if statement
# intuitive and easy to read, but provides no feedback if item is not present
if 'staghorn' in ferns:	
   i = ferns.index('staghorn')

# idiomatic approach using conditional assignment
# compact and provides feedback (-1 is a common error code)
i = ferns.index('cow') if 'cow' in ferns else -1
```

Python offers several very different ways to **remove** items from a list. For common approaches are shown below. They are all useful, so it's worth being aware they exist.

```python
# remove last item and optionally return the value
ferns.pop()					    # removes last item; returns if assigned

# remove by position (mutating; does not return items)
del ferns[2]					# removes third item
del ferns[:2]					# removes first 3 items
del ferns[1::2]				    # removes even-indexed items

# remove by value (mutating; does not return items)
ferns.remove('maidenhair')		# only removes first match

# remove duplicates
ferns = list(set(ferns))    	# converts to set and back to list; warning, may alter order!

# remove all items
ferns.clear()				    # converts to an empty list
```

The real power of sets lies in **set operations**, which provide an elegant and computationally efficient ways to combine and filter items. Note that many set operations can be specified using either an operator or a method; some, however, only have methods.

```python
# first, define some sets to work with
A = {1, 2, 3, 4, 5, 6}
B = {2, 4, 6, 8, 10, 12}
C = {22, 33, 44, 55, 66}
```

It is important to bear in mind which of the following operations are *transitive* (order does not matter) and which are not (order matters).   

A **union** of two or more sets contains every item in every set, with any duplicates removed.

```python
# union is transitive
u = A | B					        # returns union
u = A.union(B)				        # same as above
u = A | B | C			            # union of three sets
u = A.union(B, C)		            # same as above
```

```python
u
```

An **intersection** of two or more sets contains items present in every set, with all other items removed.

```python
# intersection is transitive
i = A & B					        # returns intersection 
i = A.intersection(B)	            # same as above
```

```python
i
```

A **difference** between two sets contains items present in the first but not the second. 

```python
# difference is not transitive
d = A - B					        # returns difference
d = A.difference(B)			        # same as above
d = B - A                           # not the same as the two above
```

```python
d
```

A **symmetric difference** between two sets contains all items present only in one set. In other words, it is the difference between the union and intersection.  

```python
# symmetric difference is transitive
sd = A ^ B				            # returns difference; order matters
sd = A.symmetric_difference(B)      # same as above
```

```python
sd
```

The examples above leave the original sets unchanged and return a new object that can be assigned. It is also possible to add items to an existing set with set operations. This process uses **mutating methods** to alter the the data object that calls them (but not the data object passed to the method as an argument). 

```python
# udpating based on set operations
A.update(B)			                # updates A as union with B; B unchanged
A = {1, 2, 3, 4, 5, 6}              # reset A
A.intersection_update(B)            # updates A as intersection with B
A.symmetric_difference_update(B)    # updates A as sym diff with B
# parallel update_ methods exist for other set operations
```

```python
A
```

Finally, it is possible to test for **overlap** between two sets. Tthese operations are limited to 2 sets and are not mutating; they return bool values.

```python
# testing for overlaps among two sets  
b = A.isdisjoint(B)		            # True if no items are in common
b = A.issubset(B)			        # True if all items in A are in B
b = A.issuperset(B)		            # True if all items in B are in A
```

```python
b
```

### Working with dictionaries   
    
Dictionaries are common in Python programs. They provide several important capabilities: very fast retrieval of values based on keys rather than numeric indexing; flexible value storage; intuitive mapping of real-world objects; a way to organize values that are not easily accessed through numeric indexes.   
    
Dictionaries are composed of key:value pairs. The keys are "look up" values and are immutable; they can be strings or integers. The values are data associated with the key, and mutable to allow updates. An example is using the name of a chemical as a key, and associating this with list of several values: molecular weight, solubility, and pH in aqueous solution. Indexing the dictionary with name of a chemical will return the three stored values as a list.

```python
# run this code block before running the examples that follow
birds = ['parrot', 'quail', 'duck']
genera = ['Forpus', 'Coturnix', 'Anser']
tmnt = {'name': ['role', 'color', 'weapon'], 
        'Leonardo': ['leader', 'blue', 'katana'], 
        'Raphael' : ['muscle', 'red', 'sai'],
        'Donatello' : ['brains', 'purple', 'bo'],
        'Michelangelo' : ['comedian', 'orange', 'nunchuk']}
```

It is possible to **create** a dictionary by hand, but it is much more common to prepare keys and values in lists in advance and they zip them into a dictionary.

```python
# create a dictionary
my_dict = {'A':1, 'B':2, 'C':3}	           # tedious!
my_dict = {}					           # creates an empty dictionary (not set)

# from an iterable with enumerate
my_dict = dict(enumerate(birds))	       # create a dictionary from the list
my_dict = dict(enumerate(birds, start=1))  # starts numbering at 1

# from two iterables with zip
my_dict = dict(zip(birds, genera))         # zip returns a list of tuples
```

```python
my_dict
```

Python provides two ways to **update** values in a dictionary. (Keys are immutable and cannot be updated.) Note that both approaches will **add** a new key:value pair if the key provided does not already exist.

```python
# updating / adding values using assignment and key-based indexing
my_dict['parrot'] = 'Psittacus'	                                # updates if key exists; else adds pair

# updading / adding values using .update()
my_dict.update({'parrot':'Psittacus'})	                        # same as above

# if you want ensure that no new key:value pairs are added, first test for membership 
if 'parrot' in my_dict: my_dict['parrot'] = 'Psittacus'         # using key-based indexing
if 'parrot' in my_dict: my_dict.update({'parrot':'Psittacus'})	# using .update()
```

```python
my_dict
```

Python provides several distinct ways to **remove** items from a dictionary, each useful in certain situations. `pop()` and `popitem()` return tuples that can be assigned.

```python
# remove values from a dictionary
# based on key value
del tmnt['Raphael']			# removes if key exists; else error
tmnt.pop('Raphael')			# same as above; returns if assigned

# based on last added
tmnt.popitem()				# removes the most recently added item

# randomly
tmnt.pop()					# removes at random

# remove all items
tmnt.clear()			    # becomes an empty dictionary
```

```python
tmnt
```

Learning about the **contents** of a dictionary is mostly similar to working with a list. The usual functions and operators work. Be aware, however, of the exceptions below.

```python
# testing for membership in a dictionary
# membership tests only for keys
'Donatello' in tmnt				                   # returns True iff 'Donatello' is a key

# testing for membership in values requires an extra step
'katana' in tmnt.values()		                   # returns True iff 'katana' is the entire value
'Anser' in my_dict.values()		                   # returns True iff 'Anser' is the entire value

# testing for equality of values directly *always* returns False
my_dict.values() == tmnt.values()	               # returns False
my_dict.values() == my_dict.values()               # returns False

# testing for equality of values requires an indirect comparison 
list(my_dict.values()) == list(my_dict.values())   # returns True
```

**Retrieving** information from a dictionary is different from other data classes, since keys and values work differently. 

```python
# retrieving values from a dictionary
# accessing values based on key
my_dict['parrot']				    # returns value; error if key missing
my_dict.get('parrot')			    # same; returns None if key missing
my_dict.get('cow', 'no key!')	# same; returns custom message if key missing

# accessing contents; these methods all return special data structures; all are iterable
tmnt.items()					# returns a dict items object
tmnt.keys()					    # returns a dict keys object 
tmnt.values()				    # returns a dict values object 

# accessing contents directly; these are easier to work with after extraction
list(tmnt.items())			    # returns a list of key:value tuples
list(tmnt.keys())				# returns a list of keys
list(tmnt.values())			    # returns a nested list of values
```

**Sorting** items in a dictionary works similarly to lists. Note, however, that functions and methods for sorting operate on *keys*, not values.

```python
# sort alphabetically by key
sorted(tmnt)
```
