**f-strings** provide a syntax for formatting strings when passed to `print()`. To use f-strings, include `f` or `F` immediately after the left round bracket, immediately followed by the string. 
    
A salient feature of f-strings is that you can pass variables names, and their values will be substituted into the output string.   
    
Many formatting options are available; the following are just a few examples. Date and time formatting are not covered here, but are very useful and well worth getting to know.

```python
# values for the following examples
var1, var2, var3, var4 = 800, 2.23333252342424, 2000000000, 2000000.23242352

# examples of formatting numbers
print(f"The number is {var1}")	    # displayed in default form 
print(f"The number is {var1:+}")	# includes + before number; - is displayed by default 
print(f"The number is {var1:06}")	# left pads with zeros 
print(f"The number is {var2:.4f}")	# to 4 significant figures; rounds; note decimal
print(f"The number is {var3:,}")	# use comma to separate groups of 3 numerals
print(f"The number is {var3:_}")	# use underscore to separate groups of 3 numerals
print(f"The number is {var3:,.2f}")	# comma separator and change to float with 2 signficant figures
print(f"The number is {var4:,.4f}")	# comma separator and 4 signficant figures
print(f'The number is {var3:e}')	# converts to scientific 
```

```python
# values for the following examples
var5, var6, var7 = 'Mars', 2, 3390

# the contents of curly braces are evaluated at runtime, allowing for expressions
print(f"The planet {var5} has {var6} moons.")				
print(f"The planet {var5} has a diameter of {var7 * 2} kilometers.")
print(f"The word {var5} is {len(var5)} letters long.")
print(f"The cube of {var6} is {var6**3}.")
```
