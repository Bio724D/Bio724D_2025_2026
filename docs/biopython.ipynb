```python
from Bio import SeqIO
from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord
```

```python
records = list(SeqIO.parse("crypto_diversity_CAC1.fasta", "fasta"))

first = records[0]
first
```

```python
first_sequence = first.seq
first_sequence
```

```python
print(first_sequence.count("A"))
print(first_sequence.find("TTT"))
```

```python
first_sequence.reverse_complement()
```

```python
my_rna = first_sequence.transcribe()
my_rna
```

```python
my_protein = my_rna.translate()
my_protein
```

```python
my_protein.count("A")
```
